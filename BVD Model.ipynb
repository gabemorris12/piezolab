```python
import sympy as sp

Lm, Cm, Rm, CE, s = sp.symbols('L_m C_m R_m C_E s', real=True, positive=True)

Z1 = 1/(s*Cm) + Rm + s*Lm
Z2 = 1/(s*CE)
expr = 1/(1/Z1 + 1/Z2)
expr = expr.simplify()
expr
```

```python
num, den = sp.fraction(expr)
num.expand()
```

```python
den.expand()
```

```python
omega = sp.Symbol(r'\omega', positive=True, real=True)
Z = expr.subs(s, sp.I*omega)
Z.simplify()
```

```python
Z_abs = sp.Abs(Z)
Z_abs.simplify()**2
```

```python
print(sp.latex(Z_abs.simplify()**2))
```

```python
Z_abs.simplify()
```

```python
Z_abs_lamb = sp.lambdify((Lm, Cm, Rm, CE, omega), Z_abs.simplify())

import inspect
print(inspect.getsource(Z_abs_lamb))
```

```python
Z_real = sp.re(Z).simplify()
Z_imag = sp.im(Z).simplify()
display(Z_real)
display(Z_imag)
```
